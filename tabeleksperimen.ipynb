```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# DATA HASIL 16 EKSPERIMEN
# ============================================================

results = [
    # MFCC, K, Accuracy, Precision, Recall, F1
    [13, 3, 0.9694, 0.9734, 0.9796, 0.9765],
    [13, 5, 0.9717, 0.9778, 0.9785, 0.9782],
    [13, 7, 0.9685, 0.9693, 0.9825, 0.9759],
    [13, 9, 0.9588, 0.9629, 0.9739, 0.9684],

    [18, 3, 0.9787, 0.9829, 0.9843, 0.9836],
    [18, 5, 0.9845, 0.9872, 0.9889, 0.9880],
    [18, 7, 0.9822, 0.9819, 0.9907, 0.9863],
    [18, 9, 0.9831, 0.9830, 0.9911, 0.9870],

    [20, 3, 0.9808, 0.9836, 0.9868, 0.9852],
    [20, 5, 0.9849, 0.9841, 0.9928, 0.9884],
    [20, 7, 0.9847, 0.9872, 0.9893, 0.9882],
    [20, 9, 0.9849, 0.9854, 0.9914, 0.9884],

    [24, 3, 0.9891, 0.9876, 0.9957, 0.9916],
    [24, 5, 0.9880, 0.9893, 0.9921, 0.9907],
    [24, 7, 0.9854, 0.9865, 0.9911, 0.9888],
    [24, 9, 0.9838, 0.9823, 0.9928, 0.9876]
]

columns = [
    'MFCC',
    'K',
    'Accuracy',
    'Precision',
    'Recall',
    'F1-Score'
]

df_results = pd.DataFrame(results, columns=columns)

# Tampilkan tabel
print("=== TABEL HASIL 16 EKSPERIMEN ===")
print(df_results.to_string(index=False))
```

```python
df_table = df_results.copy()

for col in ['Accuracy', 'Precision', 'Recall', 'F1-Score']:
    df_table[col] = (df_table[col] * 100).round(2)

print(df_table.to_string(index=False))
```

```python
# ============================================================
# KONFIGURASI TERBAIK
# ============================================================

best_accuracy = df_results.loc[df_results['Accuracy'].idxmax()]
best_f1 = df_results.loc[df_results['F1-Score'].idxmax()]
best_precision = df_results.loc[df_results['Precision'].idxmax()]
best_recall = df_results.loc[df_results['Recall'].idxmax()]

print("=== HASIL TERBAIK ===")

print("\nAccuracy tertinggi:")
print(best_accuracy)

print("\nF1-Score tertinggi:")
print(best_f1)

print("\nPrecision tertinggi:")
print(best_precision)

print("\nRecall tertinggi:")
print(best_recall)
```

```python
# ============================================================
# GRAFIK ACCURACY
# ============================================================

plt.figure(figsize=(10, 6))

for mfcc in sorted(df_results['MFCC'].unique()):
    
    data = df_results[df_results['MFCC'] == mfcc]
    
    plt.plot(
        data['K'],
        data['Accuracy'] * 100,
        marker='o',
        label=f'MFCC {mfcc}'
    )

plt.xlabel('Nilai K')
plt.ylabel('Accuracy (%)')
plt.title('Perbandingan Accuracy Berdasarkan MFCC dan K')
plt.xticks([3, 5, 7, 9])
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# GRAFIK F1-SCORE
# ============================================================

plt.figure(figsize=(10, 6))

for mfcc in sorted(df_results['MFCC'].unique()):
    
    data = df_results[df_results['MFCC'] == mfcc]
    
    plt.plot(
        data['K'],
        data['F1-Score'] * 100,
        marker='o',
        label=f'MFCC {mfcc}'
    )

plt.xlabel('Nilai K')
plt.ylabel('F1-Score (%)')
plt.title('Perbandingan F1-Score Berdasarkan MFCC dan K')
plt.xticks([3, 5, 7, 9])
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# GRAFIK SEMUA METRIK
# ============================================================

best = df_results.loc[df_results['F1-Score'].idxmax()]

metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
values = [best[m] * 100 for m in metrics]

plt.figure(figsize=(8, 5))

plt.bar(metrics, values)

plt.ylim(90, 100)
plt.ylabel('Score (%)')
plt.title(
    f'Performa Konfigurasi Terbaik '
    f'(MFCC={int(best["MFCC"])}, K={int(best["K"])})'
)

plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
