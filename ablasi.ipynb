```python
import pandas as pd
import numpy as np
import librosa
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import warnings
warnings.filterwarnings('ignore')

# ======================================================
# 1. EKSTRAKSI MFCC (SELURUH DATA, SEPERTI KODE BIASA)
# ======================================================
metadata_path = 'validated.tsv'
audio_folder = 'clips'

df = pd.read_csv(metadata_path, sep='\t')
df_filtered = df[df['gender'].isin(['male_masculine', 'female_feminine'])].copy()
print(f"Total data setelah filter: {len(df_filtered)}")

def extract_mfcc(file_path, n_mfcc=24):
    try:
        y, sr = librosa.load(file_path, sr=None)
        mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=n_mfcc)
        return np.mean(mfcc, axis=1)
    except:
        return None

# Ekstrak fitur
X_list = []
y_list = []
for _, row in df_filtered.iterrows():
    audio_path = os.path.join(audio_folder, row['path'])
    feat = extract_mfcc(audio_path, n_mfcc=24)
    if feat is not None:
        X_list.append(feat)
        y_list.append(row['gender'])

X = np.array(X_list)
labels = np.array(y_list)
print(f"✅ X shape: {X.shape}, labels: {len(labels)}")

# ======================================================
# 2. ABLASI DENGAN STANDARDISASI (seperti kode biasa)
# ======================================================
top_models = [
    {'mfcc': 24, 'k': 7, 'label': 'M24_K7'},
    {'mfcc': 24, 'k': 3, 'label': 'M24_K3'},
    {'mfcc': 20, 'k': 3, 'label': 'M20_K3'}
]

train_sizes = [0.6, 0.7, 0.8, 0.9]  # 60:40, 70:30, 80:20, 90:10
results_ablation = []

for model in top_models:
    mfcc_count = model['mfcc']
    k = model['k']
    model_name = model['label']
    
    X_model = X[:, :mfcc_count]  # ambil sejumlah koefisien MFCC
    
    for train_size in train_sizes:
        test_size = 1 - train_size
        
        X_train, X_test, y_train, y_test = train_test_split(
            X_model, labels,
            test_size=test_size,
            random_state=42,
            stratify=labels
        )
        
        # StandardScaler (sama seperti kode biasa)
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)
        
        knn = KNeighborsClassifier(n_neighbors=k, metric='euclidean')
        knn.fit(X_train_scaled, y_train)
        y_pred = knn.predict(X_test_scaled)
        
        acc = accuracy_score(y_test, y_pred)
        prec = precision_score(y_test, y_pred, pos_label='female_feminine')
        rec = recall_score(y_test, y_pred, pos_label='female_feminine')
        f1 = f1_score(y_test, y_pred, pos_label='female_feminine')
        tn, fp, fn, tp = confusion_matrix(y_test, y_pred, labels=['female_feminine', 'male_masculine']).ravel()
        
        results_ablation.append({
            'Model': model_name,
            'MFCC': mfcc_count,
            'K': k,
            'Train_Size': f'{int(train_size*100)}%',
            'Test_Size': f'{int(test_size*100)}%',
            'Accuracy': acc,
            'Precision': prec,
            'Recall': rec,
            'F1_Score': f1,
            'TP': tp,
            'FP': fp,
            'FN': fn,
            'TN': tn
        })

df_ablation = pd.DataFrame(results_ablation)

# ======================================================
# 3. SIMPAN KE EXCEL
# ======================================================
try:
    with pd.ExcelWriter('hasil_eksperimen.xlsx', engine='openpyxl', mode='a') as writer:
        df_ablation.to_excel(writer, sheet_name='Ablasi_FullData', index=False)
    print("\n✅ Hasil ablasi ditambahkan ke sheet 'Ablasi_FullData'")
except FileNotFoundError:
    with pd.ExcelWriter('hasil_eksperimen.xlsx', engine='openpyxl') as writer:
        df_ablation.to_excel(writer, sheet_name='Ablasi_FullData', index=False)
    print("\n✅ File hasil_eksperimen.xlsx dibuat dengan sheet 'Ablasi_FullData'")

# ======================================================
# 4. TAMPILKAN RINGKASAN
# ======================================================
print("\n📊 Ringkasan Hasil Ablasi (seluruh data + StandardScaler):")
print(df_ablation[['Model', 'Train_Size', 'Accuracy', 'F1_Score']].to_string(index=False))

print("\n📈 Statistik per Model:")
summary = df_ablation.groupby('Model').agg({
    'Accuracy': ['mean', 'std'],
    'F1_Score': ['mean', 'std']
}).round(4)
print(summary)

# Tampilkan hasil untuk split 80:20 sebagai pembanding
print("\n🔍 Hasil untuk split 80:20 (sama dengan eksperimen awal):")
print(df_ablation[df_ablation['Train_Size']=='80%'][['Model', 'Accuracy', 'F1_Score']].to_string(index=False))
```

```python
import os
from collections import defaultdict, Counter
import pandas as pd
import librosa

# load metadata
df = pd.read_csv("validated.tsv", sep="\t")

# filter hanya male dan female
df = df[df["gender"].isin(["male_masculine", "female_feminine"])].copy()

print("Jumlah per gender:")
print(df["gender"].value_counts())

print(f"Total data yang diproses: {len(df)}")

sr_by_gender = defaultdict(Counter)
failed = 0

for _, row in df.iterrows():
    audio_path = os.path.join("clips", row["path"])
    try:
        # sr=None agar sample rate asli file dipertahankan
        _, sr = librosa.load(audio_path, sr=None, mono=False)
        sr_by_gender[row["gender"]][sr] += 1
    except Exception:
        failed += 1

print("\nDistribusi sample rate asli per gender:")
for gender in ["male_masculine", "female_feminine"]:
    print(f"\n{gender}:")
    for sr, count in sorted(sr_by_gender[gender].items()):
        print(f"  {sr}: {count}")

print(f"\nJumlah file gagal dibaca: {failed}")
```
